# 2 DICE PROBABILITY EXAMPLE

```python
import pandas as pd
import random
```

```python
#local variable setup
No_of_dice = 10
Trials = 500
```

```python
combinations = 6**No_of_dice
print(combinations)
```

```python
#lets create the boundary 
TRIALS = []
```

Lets create the iterations 

```python
for i in range(Trials):
    roll_the_dice = [random.randint(1,6) for j in range(No_of_dice)]
    TRIALS.append(roll_the_dice)

roll_the_dice  
```

```python
Dice_data = pd.DataFrame(TRIALS,
                         columns=[f"Dice {i}" for i in range(1,No_of_dice + 1)]
                        )
Dice_data["Sum"] = Dice_data.sum(axis=1)

Count_7 = (Dice_data["Sum"] == 32).sum()

probability_trials = Count_7 / Trials
```
